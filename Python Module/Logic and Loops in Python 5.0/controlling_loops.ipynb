### Exercise One: Marine species Counter
#### Using a for loop to count the number of endangered marine species in a list

```python
endangered_marine_species = ['Hawksbill Turtle', 'Vaquita', 'Blue Whale', 'Staghorn Coral', 'Green Turtle']

endangered_count = 0

for species in endangered_marine_species:
    endangered_count += 1

print(f"Number of endangered marine species: {endangered_count}")
```

### Exercise 2: Deforestation alert
#### Use a while loop to simulate a deforestation monitoring system that alerts when the forested area falls below a critical threshold

```python
initial_forested_area = 1000    # in square kilometers
deforestation_rate = 20        # square kilometers per year

years = 0
while initial_forested_area > 500:
    initial_forested_area -= deforestation_rate
    years += 1

print(f"Years until critical deforestation level: {years}")
```

### Exercise 3: Biodiversity Analysis
#### Using a for loop to analyse a list of animals, printing out whether they are endangered or not

```python
animals = ['Tiger', 'Blue Whale', 'African Elephant', 'Koala', 'Panda']
endangered_animals = ['Tiger', 'Blue Whale', 'African Elephant']

for animal in animals:
    if animal in endangered_animals:
        print(f"{animal} endangered")
```

### Exercise 4: Ocean pollution tracker

### Exercise 5: Species conservation
#### Implementing a while loop to simulate a conservation effort, stopping when a certain population target is reached

```python
current_population = 150
target_population = 500
years = 0
growth_rate = 1.07

while current_population < target_population:
    current_population *= growth_rate
    years += 1

print(f"Years to reach target population: {years}")
```

### Exercise 6: Basic List comprehension
#### Creating a list of squares of numbers from 1 to 10 using list comprehension

```python
numbers = range(1, 11)

squares = [x ** 2 for x in numbers]

print(squares)
```

### Exercise 7: List Comprehension with a conditional
#### Using List comprehension to create a list of only even numbers from 1 to 20

```python
numbers = range(1, 21)

even_numbers = [x for x in numbers if x % 2 == 0]

print(even_numbers)
```

### Exercise 8: Dictionary comprehension
#### Creating a Dictionary that categorises each marine animal in the list as 'Aquatic' using Dictionary Comprehension

```python
marine_animals = ['Dolphin', 'Shark', 'Turtle']

animal_categories = {animal: 'Aquatic' for animal in marine_animals}

print(animal_categories)
```

### Exercise 9: Data mapping with dictionary comprehension
#### Using dictionary comprehension to create a mapping of each marine species to its conservation status based on the given lists, only including species marked as endangered

```python
species = ['Dolphins', 'Whales', 'Sea Turtles', 'Seals']
status = ['Endangered', 'Vulnerable', 'Endangered', 'Least Concern']

number_of_species = range(len(species))

endangered_species = {species[x]: status[x] for x in number_of_species if x % 2 == 0}

print(endangered_species)
```

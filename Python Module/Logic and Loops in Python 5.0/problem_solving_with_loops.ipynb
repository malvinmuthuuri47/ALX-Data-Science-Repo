### Exercise One: Ocean cleanup analysis
#### Rearrange the code blocks to create a program that calculates the total amount of waste collected in an ocean cleanup campaign over a week

```python
weekly_collection = [340, 220, 455, 300, 385, 520, 410]

total_waste_collected = 0

for daily_collection in weekly_collection:
    total_waste_collected += daily_collection

print(f"Total waste collected over the week: {total_waste_collected} kg")
```

### Exercise Two: Endangered species monitoring
#### Rearrange the code to create a program that tracks sightings of endangered species over a month, stopping if a certain species is spotted

```python
sightings = ['Deer', 'Tiger', 'Blue Whale', 'Mountain Gorilla', 'Elephant']
endangered_species = ['Tiger', 'Blue Whale', 'Mountain Gorilla']

for species in sightings:
    if species in endangered_species:
        print(f"Endangered species spotted: {species}")
        break
```

### Exercise Three: Forest growth monitoring
#### Rearrange the code blocks to create a program that simulates the growth of a forest over years until it reaches a specified threshold area. The forest growth is compounded annually.

```python
forest_area = 1000
growth_rate = 1.05

year = 0

while forest_area < 1500:
    forest_area *= growth_rate
    year += 1
    print(f"Forest area after {year} years: {forest_area:.2f} sq km")
```

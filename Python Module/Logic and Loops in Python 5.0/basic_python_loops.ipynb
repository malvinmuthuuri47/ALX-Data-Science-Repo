## This notebook demonstrates how to use Python Loops to solve some common Programming issues

### Task One: Using a Python For Loop to print an iterable's content

```python
animals = ['Great White Shark', 'Blue Whale', 'African Elephant', 'Bald Eagle', 'Orangutan', 'Tiger', 'Panda', 'Koala']

for animal in animals:
    print(animal)
```

### Task Two: Using a While loop with a condition

```python
i = 0

while i < 5:
    print(i)
    i+=1
```

### Task Three: Using break in a Loop

```python
animals = ['Great White Shark', 'Blue Whale', 'African Elephant', 'Bald Eagle', 'Orangutan', 'Tiger', 'Panda', 'Koala']

for animal in animals:
    if animal == "Orangutan":
        break
    
    print(animal)
```

### Task Four: Using Continue in a Loop

```python
animals = ['Great White Shark', 'Blue Whale', 'African Elephant', 'Bald Eagle', 'Orangutan', 'Tiger', 'Panda', 'Koala']

for animal in animals:
    if animal == 'Bald Eagle':
        continue

    print(animal)
```

### Task Five: Using pass in a loop

```python
animals = ['Great White Shark', 'Blue Whale', 'African Elephant', 'Bald Eagle', 'Orangutan', 'Tiger', 'Panda', 'Koala']

for animal in animals:
    if animal == 'Panda':
        pass

    print(animal)
```

### This notebook contains code that's meant to introudce the concept of modules and packages. It also contains a few additional examples on how modules can be used

```python
content = """
s = 'Hello '

def say_hi(name):
    print(s + name)

class Greet:
    pass
"""
```

```python
# write the above text to a file called my_module.py within our current working directory

with open('./my_module.py', 'w') as fp:
    fp.write(content)
```

```python
import my_module
```

```python
try:
    print(s)
except NameError:
    print("Variable 's' does not exist!")
```

```python
# accessing the module functions and variables correctly

my_module.s

my_module.say_hi('Nelson')
```

#### Importing modules using an alias

```python
# alias for my_module
import my_module as md
```

```python
# Accessing module variable using an alias
md.s
```

```python
# Accessing module functions using an alias
md.say_hi('Jabulani')
```

```python
# Accessing the empty class, Greet
md.Greet
```

#### Importing Modules directly

```python
from my_module import s, say_hi, Greet
```

```python
s
```

```python
say_hi('Jabulani')
```

```python
Greet
```

```python
# listing all the functions names in a module

import math

# use the sqrt function in the math module
x = math.sqrt(81)
print('The square root of 81 is equal to {}'.format(x))

# list all functions in math module
list_all = dir(math)
print('Functions in the math module: {}'.format(list_all))
```

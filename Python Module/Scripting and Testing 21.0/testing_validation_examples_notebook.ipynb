## Testing Methods

```python
def range_checker(x):
    """
    Function that accepts only integers from 0 to 100, both 0 and 100,
    both 0 and 100 included.
    """
    if x < 1 or x > 100:
        print("Error")
    else:
        print("Success")
```

### Equivalence Classes

```python
a = -50
b = 50
c = 110

range_checker(a)
range_checker(b)
range_checker(c)
```

### Boundary Values

```python
a = 0
b = 2
c = 99
d = 101

range_checker(a)
range_checker(b)
range_checker(c)
range_checker(d)
```

### Path Testing

```python
a = 0
b = 30

def decision(a, b):
    if a < 25:
        print("error in a")
    else:
        print("no error in a")
    if b < 25:
        print("error in b")
    else:
        print("no error in b")

decision(a, b)
```

## Finding Errors

### Tracing

#### Tracing refers to inserting temporary statements into code to output values at various stages of its execution. This practice can be extremely useful when there is no debugger.

```python
y = 2
x = y * y * 2
z = x + 5

print(f'Z equals: {z}')  # <-- Trace instruction. The program will output the current version of z at this point
if z == 13:    # before it is evaluated by the conditional if statement
    print('I got here')  # <-- Trace instruction. Used to check if we entered the conditional statement correctly.
```

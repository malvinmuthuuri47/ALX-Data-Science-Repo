### This notebook contains code examples that are meant to help the learner understand the basic functions of Pandas

```python
import pandas as pd
import numpy as np
```

#### Exercise 1: Creating a Pandas DataFrame from a dictionary.

```python
data = {
	"Crop": ['Wheat', 'Corn', 'Rice', 'Soybean'],
	"Yield_per_Acre": [3, 4.5, 5, 2.8],
	'Country': ['USA', 'China', 'India', 'Brazil']
}

df = pd.DataFrame(data)
```

#### Exercise 2: Accessing DataFrame observations for index values 1 and 2 from the dataset

```python
df_observations = df.iloc[1:3]

print(df_observations)
```

```python
type(df_observations)
```

#### Exercise 3: Create a 3x2 array using NumPy linrary to store information for different tree species

```python
tree_array = np.array([[30, 25], [60, 30], [25, 20]])

df = pd.DataFrame(tree_array, index=["Maple", "Pine", "Birch"], columns=["Max_Height_m", "Growth_Rate_cm_per_year"])

print(df)
```

#### Exercise 4: Display the Max_Height_m and Growth_Rate_cm_per_year columns for the Pine Tree

```python
extracted_data = df.loc["Pine"]

print(extracted_data)
```

#### Exercise 5: Loading a CSV file using Pandas and creating a DataFrame from the loaded data

```python
data = pd.read_csv('https://raw.githubusercontent.com/Explore-AI/Public-Data/master/Data/Python/Animals.csv')

animals_df = pd.DataFrame(data)

print(animals_df.head())
```

#### Exercise 6: Accessing and displaying information on animals listed as critically endangered in the conservation status column

```python
critically_endangered = animals_df[animals_df["Conservation Status"] == "Critically Endangered"]

print(critically_endangered)
```

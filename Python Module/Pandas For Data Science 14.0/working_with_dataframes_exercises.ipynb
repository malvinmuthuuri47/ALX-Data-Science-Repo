### This notebook contains the code exercises meant to help the learner understand and learn how to work with DataFrames

```python
import pandas as pd
import numpy as np
```

```python
animal_df = pd.read_csv('https://raw.githubusercontent.com/Explore-AI/Public-Data/master/Data/Python/Animals.csv')
```

#### Exercise 1: Filter the dataset to show only species that have an average lifespan (Years) greater than 20 and are classified as vulnerable in conservation status

```python
animal_df.head()
```

```python
avg_lf_animals_df = animal_df[(animal_df["Average Lifespan (Years)"] > 20) & (animal_df["Conservation Status"] == "Vulnerable")]

print(avg_lf_animals_df)
```

#### Exercise 2: Creating a new column and deleting it from a DataFrame

```python
# Creating the Column
animal_df['Lifespan in Months'] = animal_df['Average Lifespan (Years)'] * 12

animal_df.head()
```

```python
# Dropping the column
animal_df.drop('Lifespan in Months', axis=1).head()
```

#### Exercise 3: Filtering Dataframes and suing apply and lambda on Dataframes

```python
# Filter the data
long_lived_species = animal_df[animal_df['Average Lifespan (Years)'] > 50]

long_lived_species
```

```python
# Convert Average Lifespan (Years) of all species into the float data type in long_lived_species

long_lived_species.info()
```

```python
long_lived_species['Average Lifespan (Years)'] = long_lived_species['Average Lifespan (Years)'].apply(lambda x: float(x))

long_lived_species.info()
```

#### Exercise 4: Listing all columns for species that have the string "endangered" as part of their conservation status

```python
animal_df.head()
```

```python
# animal_df[animal_df['Conservation Status'] == 'Endangered']

# Filtering to find species with 'Endangered' in their conservation status
Endangered_species = animal_df[animal_df['Conservation Status'].str.contains('Endangered')]

# Display the filtered DataFrame
Endangered_species
```

#### Exercise 5: Listing the average lifespan of species for each habitat in descending order

```python
animal_df.groupby('Habitat')['Average Lifespan (Years)'].mean().sort_values(ascending=False)
```

```python
animal_df.head()
```

#### Exercise 6: Creating a column that categorizes species into different classes based on their conservation status

```python
animal_df['Threat Level'] = animal_df['Conservation Status'].apply(
    lambda status:
        'High' if status in ['Critically Endangered', 'Endangered'] else
        'Medium' if status in ['Vulnerable', 'Near Threatened'] else
        'Low'
)

animal_df.head()
```

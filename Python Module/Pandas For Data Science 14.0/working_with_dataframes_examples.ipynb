## This notebook walks the learner throug the process of manipulating, transforming, and altering Pandas DataFrames

```python
import pandas as pd
```

```python
# load data - pass 'Name' as our index column
df = pd.read_csv('https://raw.githubusercontent.com/Explore-AI/Public-Data/master/Data/fundamentals/football_players.csv', index_col='Name')

# use the head() function to look at the first 5 rows
df.head()
```

```python
df[df['Age'] > 30]
```

#### Creating new Columns from existing columns in a Pandas DataFrame

```python
# Create a column of rating per yaer of age
df['Rating Per Year of Age'] = df['Overall'] / df['Age']

df['Rating Per Year of Age'].head()
```

#### Deleting Columns using the drop() function

```python
# Drop the column we just created above
df = df.drop('Rating Per Year of Age', axis=1)

df.head()
```

```python
# Look at the average rating by age and nationaloty (first 15 rows)
df.groupby(['Age', 'Nationality'])[['Overall']].mean().head(15)
```

## Transforming Data

```python
# Applying a custom function to a dataframe
def year_to_month(x):
    "converts no of years to no of months"
    return x * 12
```

```python
# Change the age column to months and look at the first 5 entries
df['Age'].apply(year_to_month).head()
```

```python
# Using a lambda function on a DataFrame
year_to_month_lamb = lambda x: x * 12
```

```python
# print function answer
print('def: ', year_to_month(30))
print('lambda: ', year_to_month_lamb(30))
```

```python
# Change the age column to months and look at the first 5 entries
df['Age'].apply(lambda x: x * 12).head()
```

```python
# Create Function
def position_type(s):
    """
    This function converts the individual positions (abbreviations) and classifies it as
    either a forward, midfielder, Back, or Goalkeeper
    """
    if (s[-2] == 'T') | (s[-2] == 'W'):
        return 'Forward'
    elif s[-2] == 'M':
        return 'MidFielder'
    elif s[-2] == 'B':
        return 'Back'
    else:
        return 'GoalKeeper'
```

```python
# Create position type column
df['Preferred Positions Type'] = df['Preferred Positions'].apply(position_type)

# Look at the first 5 entries
df['Preferred Positions Type'].head()
```

## Transforming column types

```python
# Select all numeric attribute columns, i.e. excluding "word-type" columns such as Nationality
cols = ['Overall', 'Acceleration', 'Aggression', 'Agility', 'Balance',
        'Ball control', 'Composure', 'Crossing', 'Curve', 'Dribbling',
        'Finishing', 'Free kick accuracy', 'GK diving', 'GK handling',
        'GK kicking', 'GK positioning', 'GK reflexes', 'Heading accuracy',
        'Interceptions', 'Jumping', 'Long passing', 'Long shots', 'Marking',
        'Penalties', 'Positioning', 'Reactions', 'Short passing', 'Shot power',
        'Sliding tackle', 'Sprint speed', 'Stamina', 'Standing tackle', 'Strength',
        'Vision', 'Volleys']
```

```python
# Create function
def to_float(x):
    """Transforms numeric attribute columns to type float"""
    if type(x) is int:
        return float(x)
    else:
        return float(x[0:2])
```

```python
# use the map() function to transform all selected columns
df[cols] = df[cols].map(to_float)
```

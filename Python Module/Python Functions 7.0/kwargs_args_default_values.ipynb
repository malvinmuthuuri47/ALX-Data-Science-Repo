#### Exercise One: Write a Python function that takes two parameters, calculating and returning the total area covered by the given number of trees

```python
def calculate_forest_area(trees, area_per_tree=10, deforestation_factor=0.2):
    """
    Calculates and returns the total area covered by the given number of trees

    Parameters:
    - trees: number of trees
    - area_per_tree: area covered by each tree
    - deforestation_factor: Percentage of the total area that has been deforested

    Returns:
    - The net forest area after considering deforestation
    """
    return trees * area_per_tree * deforestation_factor

net_forest_area = calculate_forest_area(50, 5, 0.3)

print(net_forest_area)
```

#### Exercise Two: Write a Python function that tracks the progress of tree planting campaigns

```python
def tree_planting_campaign(initial_trees, *trees_per_campaign):
    """
    Calculates the total number of trees after a variable number of tree planting campaigns

    Parameters:
    - initial_trees: The initial number of trees
    - trees_per_campaign: The total number of trees planted in each additional campaign

    Returns:
    - The total number of trees after the campaigns
    """
    total_trees = initial_trees

    for new_trees in trees_per_campaign:
        total_trees += new_trees

    return total_trees

tree_planting_campaign(100, 20, 30, 50)
```

#### Exercise One: Calculating the total carbon emission over the years using built-in functions given a list of deforested land in hectares over five years

```python
# deforested land for each year
deforested_land = [10, 15, 8, 12, 20]

# carbon emission factor (tons of CO2 per hectare)
carbon_emission_factor = 40

total_deforested_land = sum(deforested_land)

total_carbon_emission = total_deforested_land * carbon_emission_factor

print(f"Total deforested land is {total_deforested_land} hectares and total carbon emission is {total_carbon_emission}")
```

#### Exercise Two: Creating a function that accepts a parameter of the area of a forest in acres and returns the corresponding area in hectares by multiplying the acres by a conversion rate

```python
def acres_to_hectares(area_of_forest: float | int) -> str:
    '''
    Convert the area of a forest in acres to hectares by multiplying the area in acres with a conversion rate

    Parameters:
    area_of_forest: The forest area in acres

    Return:
    The area of a forest in hectares
    '''
    conversion_rate = 0.404686
    forest_size_in_hectares = area_of_forest * conversion_rate

    return f"The area of the forest in hectares is {forest_size_in_hectares}"

acres_to_hectares(50)
```

#### Exercise Three: Calculating the average deforested land using the sum and len built-in functions

```python
# deforested land for each year
deforested_land = [10, 15, 8, 12, 20]

average_deforested_land = sum(deforested_land) / len(deforested_land)

print(f"The average deforested land is {average_deforested_land}")
```

#### Exercise One: Implement a Python function that analyses the overall carbon impact considering deforestation, tree planting, and carbon offset. The function returns one of three categories, representing the overall carbon impact

```python
def analyse_carbon_impact(deforested_areas: list, planted_trees: list, carbon_emission_factor: float, carbon_offset_per_tree: float) -> str:
    """
    Calculate and categorise the overall carbon impact, considering deforestation, tree planting, and offset.

    Parameters:
    - deforested_areas (list): A list of deforested areas in hectares for each region
    - planted_trees (list): A list of trees planted in each region
    - carbon_emission_factor (float): carbon emission factor in tons of CO2 per hectare
    - carbon_offset_per_tree (float): carbon offset per tree in tons of CO2

    Returns:
    - str: 'Positive', 'Neutral', or 'Negative' based on the overall carbon impact
    """
    total_emissions = 0

    for area in deforested_areas:
        emission_per_area = carbon_emission_factor * area
        total_emissions += emission_per_area

    total_offset = 0
    for tree in planted_trees:
        offset_per_tree = carbon_offset_per_tree * tree
        total_offset += offset_per_tree

    overall_impact = total_emissions - total_offset

    if total_offset >= total_emissions:
        overall_impact = 'Positive'
    elif total_offset == total_emissions:
        overall_impact = 'Neutral'
    elif total_offset < total_emissions:
        overall_impact = 'Negative'
    else:
        overall_impact = 'An error occurred'

    return overall_impact

analyse_carbon_impact([10, 15, 8, 12, 20], [30, 40, 25, 35, 50], 30, 5)
```

#### Exercise Two: This exercise needs us to project the future of planted trees based on an initial planting, an annual growth rate, and the number of years for projection.

```python
def project_future_tree_planting(initial_planted_trees: int, annual_growth_rate: float, projection_years: int) -> list[int]:
    """
    Project the future number of planted trees for each year (starting at 1) based on the initial planting and growth rate.

    Parameters:
    - initial_planted_trees (int): Initial number of trees planted.
    - annual_growth_rate (float): Annual growth rate of planted trees.
    - projection_years (int): Number of years for projection.

    Returns:
    - list: list of projected tree planting for each year
    """
    tree_projection = [initial_planted_trees * (1 + annual_growth_rate) ** year for year in range(1, projection_years+1)]
    return tree_projection

project_future_tree_planting(100, 0.05, 10)
```

#### Exercise Three: This exercise requires the learner to calculate dynamic carbon offsets for different regions dynamically based on the provided factors by averaging their values.

```python
def calculate_region_offset_with_kwargs(region_name: str, **factors) -> tuple[str, int | float]:
    """
    Calculate dynamic carbon offset for a specified region based on various factors provided as keyword arguments.

    Parameters:
    - region_name (str): Name of the region
    - **factors: Variable number of keyword arguments.

    Returns:
    - tuple: Region name and calculated dynamic carbon offet.
    """
    sum_of_factors = 0
    
    for k,v in factors.items():
        sum_of_factors += v

    average_carbon_offset = sum_of_factors / len(factors)

    return region_name, average_carbon_offset

calculate_region_offset_with_kwargs("Amazon", deforested_area=30, planted_trees=40, carbon_emission_factor=10)
```

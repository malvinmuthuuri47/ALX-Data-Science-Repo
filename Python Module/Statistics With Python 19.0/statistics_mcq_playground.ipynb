```python
import numpy as np
from scipy import stats

student_scores = np.array([55, 62, 67, 58, 63, 59, 61, 68, 60, 64])
mu = 60
x_bar = np.mean(student_scores)
s = np.std(student_scores, ddof=1)
n = len(student_scores)
alpha = 0.05

t_stat = (x_bar - mu) / (s / np.sqrt(n))
p_value = stats.t.cdf(t_stat, df=n-1) if t_stat < 0 else (1 - stats.t.cdf(t_stat, df=n-1))
p_value *= 2

print("T Statistic: ", t_stat)
print("P-value: ", p_value)
```

### Poisson Distribution

##### P(X = k) = (e^-λ × λ^k) / k!

##### Where:
##### λ (lambda) → the average number of events in the interval
##### k          → the exact number of events you want the probability for
##### e          → Euler's number (approximately 2.71828)
##### k!         → k factorial (k × k-1 × k-2 ... × 1)

```python
import numpy as np

lambda_ = 8
k = 5
e_to_neg_lambda = np.exp(-lambda_)
lambda_to_k = lambda_ ** k
k_factorial = np.math.factorial(k)

probability = (e_to_neg_lambda * lambda_to_k) / k_factorial
print(probability)
```

```python
rate = 8
k = 6

e_to_neg_lambda = np.exp(-rate)
lambda_to_k = rate ** k
k_factorial = np.math.factorial(k)

probability = (e_to_neg_lambda * lambda_to_k) / k_factorial
print(probability)
```

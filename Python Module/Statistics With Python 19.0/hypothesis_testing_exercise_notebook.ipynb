#### Exercise 2: Write a function that calculates the test statistic (z-value) based on the sample mean, population mean, population standard deviation, and sample size.

#### The test statistic can be calculates using the formula for a one-sample z-test:
##### ***z = (x̄ - μ) / (σ / √n)***
##### where:
##### - x̄ is the sample mean
##### - μ is the population mean under the null hypothesis
##### - σ is the population standard deviation
##### - n is the sample size

```python
def test_stat_z_val(x̄, μ, σ, n):
    population_mean = μ
    population_std_dev = σ
    sample_mean = x̄
    sample_size = n
    
    return (x̄ - μ) / (σ / np.sqrt(n))
```

#### Exercise 3: Write a function that returns the lower and upper critical z-values for a two-sided test at a specific significance level

```python
def critical_z_vals(significance_level):    
    lower = norm.ppf(significance_level / 2)
    upper = norm.ppf(1 - significance_level / 2)
    
    return lower, upper
```

#### Exercise 4: Write a function which makes a decision on whether to reject or not reject the null hypothesis based on the test statistic and critical values

```python
def hypothesis_decider(z_value, lower_z, upper_z):
    if z_value < lower_z or z_value > upper_z:
        decision = "Reject null hypothesis"
    else:
        decision = "Fail to reject null hypothesis"
    
    return decision
```

```python
# Given data
sample_mean = 145    # sample mean weight in mg
population_mean = 150    # hypothesised population mean weight in mg
population_std = 20    # population standard deviation in mg
n = 30    # sample size
alpha = 0.05    # significance level

# Calculate the z-value
z = test_stat_z_val(sample_mean, population_mean, population_std, n)

# Get critical z-value
lower_z, upper_z = critical_z_vals(alpha)

# Make a decision
decision = hypothesis_decider(z, lower_z, upper_z)

print(f"Test Statistic (z-value): {z}")
print(f"Critical z-value (lower): {lower_z}, Upper: {upper_z}")
```

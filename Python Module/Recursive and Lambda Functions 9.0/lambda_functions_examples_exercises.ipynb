#### Exercise 2: Write a lambda function named sort that takes a sinble argument: a list of numbers and returns a new list with the numbers sorted in ascending order

```python
sort = lambda x: sorted(x)
```

```python
sort([6,2,3,9,1,5])
```

#### Exercise 3: Write a lambda function named filtered_list that utilizes filter() to print a list of integers that are greater than 9

```python
list1 = [2,6,8,10,11,4,12,7,13,17,0,3,21]

filtered_list = list(filter(lambda x: x > 9, list1))

print(filtered_list)
```

#### Exercise 4: Write a Lambda function named mapped_list that makes use of the map() function to output the remainder after dividing each number in a list by 2

```python
list2 = [2,6,8,10,11,4,12,7,13,17,0,3,21]

mapped_list = list(map(lambda x: x % 2, list2))

print(mapped_list)
```

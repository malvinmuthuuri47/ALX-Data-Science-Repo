## This notebook introduces the concepts of recursive functions to the learner

#### Exercise 1: The learner's task is to create a recursive mathematical function that converts f(n) = a ^ n

```python
def exponential(a, n):
    # code goes here
    if n == 0:
        return 1
    
    return a * exponential(a, n - 1)

print(exponential(2, 3))
```

#### Exercise 2: The learner's task is to create a function that uses recursion to calculate the nth number in the Fibonacci sequence

```python
def fibonacci(n):
    # code goes here
    if n == 0:
        return 0

    if n == 1:
        return 1

    return fibonacci(n - 1) + fibonacci(n - 2)

fibonacci(7)
```

#### Exercise 3: The learner's task is to write a recursive function that calculates the cumulative sum of all positive integers up to a given number n.

```python
def cumulate(n):
    # code goes here
    if n <= 0:
        return 0
    elif n == 1:
        return 1

    return n + cumulate(n - 1)

cumulate(7)
```

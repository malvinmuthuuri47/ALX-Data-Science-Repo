### This notebook contains code that's meant to showcase and introduce the learner to Numpy and some of its basic operations.

```python
import numpy as np
```

```python
# create an array by passing in a list of lists
ratings = np.array([[94, 89, 63, 45], [93, 92, 48, 23], [92, 94, 56, 98]])
```

```python
# inspecting the shape of the array
ratings.shape
```

```python
# Create an array of ones - pass in shape as a tuple
np.ones((3, 4))
```

```python
# Create an array of zeros - pass in shape as a tuple
np.zeros((2, 3))
```

```python
# Create an array of random values - pass in shape as a tuple
np.random.random((3, 3))
```

### Accessing NumPy arrays

```python
ratings
```

```python
# selecting the top left element
ratings[0, 0]
```

```python
# Selecting the first row
ratings[0, :]
```

```python
# Select the first column
ratings[:, 0]
```

```python
# Select the first two rows and first two columns
ratings[0:2, 0:2]
```

#### Boolean Array Indexing

```python
# Select all values greater than 90
ratings[ratings > 90]
```

### Modifying NumPy arrays

```python
# Append an extra row
ratings_extra_row = np.append(ratings, [[92, 88, 78, 55]], axis=0)

ratings_extra_row
```

```python
ratings
```

```python
# Delete the 3rd row
ratings_del_row = np.delete(ratings, [2], axis=0)

ratings_del_row
```

### Functions on NumPy arrays

```python
# Sum of all elements in the array
ratings.sum()
```

```python
# Sum of each row
ratings.sum(axis=1)
```

```python
# Sum of each column
ratings.sum(axis=0)
```

```python
# Min of each row
ratings.min(axis=1)
```

```python
# Finds the index of the maximum value in the entire array
overall_max_index = np.argmax(ratings)

overall_max_index
```

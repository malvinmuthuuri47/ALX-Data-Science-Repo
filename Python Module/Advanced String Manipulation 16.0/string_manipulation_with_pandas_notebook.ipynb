### This notebook contains code examples meant to assist the learner understand how to apply Pandas to manipulate strings within DataFrames and Series

#### Example 1: Suppose we have been given a dataset of different plant species recorded in a national park. The species names are inconsistently formatted with leading/trailing spaces and varying case letters.

**Clean the Species by column by removing leading/trailing spaces and converting all names to lowercase**

```python
import pandas as pd

# sample data
data = {'species': [' Maple (10 years) ', 'oak', 'Pine(3 years)', 'maple', ' Oak (1.5 Years)']}
df = pd.DataFrame(data)

# Cleaning the 'species' column
df['species'] = df['species'].str.strip().str.lower()
print(df)
```

#### In our dataset, some species have their age mentioned in years within the name in parenthesis (e.g. "Maple (10 years)").

**Use a regex to extract the age in years from the species name and create a new column 'age'. Fill with 'Unknown' if there is no age present.**

```python
df['age'] = df['species'].str.extract(r'(\d+ \w+)').fillna('Unknown')

print(df)
```

#### After cleaning the dataset, we want to know how many unique species are recorded and their respective counts

**Aggregate the data to count the number of occurrences of each unique species**

```python
df['species'] = df['species'].str.extract('([a-zA-Z]+)', expand=False).fillna('Unknown').str.strip().str.lower()

species_counts = df['species'].value_counts()
print(species_counts)
```

#### The Park authorities are interested in grouping species by their first letter to design different zones in the park.

**Create a new column 'zone' based on thefirst letter of each species name and group the dataframe by this new column**

```python
df['zone'] = df['species'].str.extract(r'(\w{1})')

print(df)
```

#### Example 1: Given a paragraph about conservation efforts, split the text into individual sentences using regular expressions

```python
import re

text = "Conservation efforts are increasing. Habitats are being restored. Species are recovering."
sentences = re.split(r"\. ", text)

print(sentences)
```

#### Example 2: Extract all numbers followed by "acres" to find references to land araa in a text. Utilize **re.compile** to create a regex pattern that matches this format

```python
import re

text = "The national park covers 5000 acres, while the community forest spans 750 acres."

# my solution
pattern = r"(\b\d{4} \w{5}\b)"
pattern_02 = r"(\b\d{3} \w{5}\b)"

compiled_pattern = re.compile(pattern)
compiled_pattern02 = re.compile(pattern_02)

answer = re.search(compiled_pattern, text)
answer2 = re.search(compiled_pattern02, text)

# Examiner's solution
examiner_pattern = re.compile(r'\d+\s*acres', re.IGNORECASE)

land_areas = examiner_pattern.findall(text)

print(answer.group())
print(answer2.group())

print(land_areas)
```

#### Example 3: Given a text with various animal and plant species names formatted as 'Genus Species', compile a regex object to find all occurrences of these species' names in the text

```python
import re

text = "In the Amazon rainforest, species like Panthera onca, Inia geoffrensis, and Enterpe precatoria are found."

pattern = re.compile(r'\b[A-Z][a-z]* [a-z]{4,}\b')

# species_names = pattern.findall(text)
species_names = [res for res in pattern.findall(text) if "rainforest" not in res]

print(species_names)
```

#### Example 4: Given a text containing different plant names related to sustainable land management, extract all occurrences of specific plants. The names to extract are "aok", "maple", and "pine". Use re.compile to optimise the pattern matching

```python
import re

text = "The forest had a variety of trees including oak, maple, and pine. Other species included birch and spruce."

# student regex
# pattern = re.compile(r'\w+')

# Examiner's regex
pattern = re.compile(r'\boak\b|\bmaple\b|\bpine\b|\bbirch\b|\bspruce\b')

species_names = pattern.findall(text)

print(species_names)
```

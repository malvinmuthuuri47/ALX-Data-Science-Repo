```python
import re
from collections import Counter
```

```python
social_media_posts = """
Great news! The GreenWood Project has successfully planted 10000 trees in the Amazon Rainforest #GreenEarth #Conservation
Update: ForestCoverApp shows a 12% increase in forest cover in the last 5 years. #TechForGood
Sad to see illegal logging in Madagascan rainforests. We need stricter laws! #SaveForests #ActNow
Celebrating World Environment Day with a pledge to plant 20000 trees. Join us! #EnvironmentDay #GoGreen
Interesting study published in NatureJournal: Rainforest biodiversity is crucial for ecological balance. #ScienceForNature
"""
```

#### Exercise 1: Write a Python function called extract_hashtags to extract all unique hashtags from the social media posts

```python
def extract_hashtags(social_media_posts: str) -> list[str]:
    """
    This function takes social media posts and extracts all the hashtags from the text
    using regular expressions

    Arguments:
        social_media_posts: A string containing the text with hashtags to be extracted

    Returns:
        A list of strings extracted that are hashtags from the social media post
    """

    pattern = r'#\w+\b'
    
    hashtags = re.findall(pattern, social_media_posts)
    return hashtags

print(extract_hashtags(social_media_posts))
```

#### Exercise 2: Write a Python function called extract_numbers to find all numbers mentioned in the posts

```python
def extract_numbers(social_media_posts: str) -> list[int | str]:
    """
    This function takes a string of social media posts and extracts numerical
    data from the posts

    Arguments:
        social_media_posts: The social media posts

    Returns:
        A list of strings with the extracted numerical and textual data
    """

    pattern = r'\d+%* \w+'
    
    nums = re.findall(pattern, social_media_posts)
    return nums

print(extract_numbers(social_media_posts))
```

#### Exercise 3: Write a function called count_specific_words to count the occurrences of the words "illegal" and "logging" in the posts

```python
def count_specific_words(social_media_posts: str) -> Counter[str]:
    """
    Counts how often words related to negative impacts ("illegal", "logging") appear

    Args:
        social_media_posts: A string composed of social media posts

    Returns:
        A counter for each occurrence of the word showing how many times it appears
    """
    pattern = r'\b\w+'

    results = re.findall(pattern, social_media_posts)
    counts = Counter(res for res in results if res in {'illegal', 'logging'})
    return counts

print(count_specific_words(social_media_posts))
```

#### Exercise 4: Write a function to extract proper names that refer to locations

```python
def extract_locations(social_media_posts: str) -> list[str]:
    """
    Extracts locations from the social_media_posts string passed to it

    Args:
        social_media_posts: social_media_posts: A string composed of social media posts

    Returns:
        A list of strings with the extracted location names
    """

    pattern = r'\b[A-Z][a-z]+ [a-z]{6,}+\b'
    
    results = re.findall(pattern, social_media_posts)
    return results

print(extract_locations(social_media_posts))
```

```python
def extract_locations(text):
    # Regular expression to match potential location names
    pattern = r'\b[A-Z][a-z]+(?:\s[A-Z][a-z]+)*(?:\s[rR]ainforests?)?\b'

    # Find all matches
    potential_locations = re.findall(pattern, text)

    # Filtering based on context and avoiding false positives
    locations = [loc for loc in potential_locations if not loc.endswith(':') and 'rainforest' in loc.lower() and loc != 'Rainforest']

    return locations

# Test with the provided text
print(extract_locations(social_media_posts))
```

#### Exercise 1. Write a bubble sort function that includes a flag that will allow the function to terminate when there is nothing left to sort

```python
def bubble_sort(arr: list[int]) -> list[int]:
    n = len(arr)

    while True:
        swapped = False

        for i in range(1, n):
            if arr[i - 1] > arr[i]:
                arr[i - 1], arr[i] = arr[i], arr[i - 1]
                swapped = True

        if not swapped:
            break
    
    return arr
```

```python
my_list = [64, 34, 25, 12, 22, 11, 90]

bubble_sort(my_list)
```

### Exercise 1: Write a basic linear seach function using the pseudocode provided below

```python
def linear_search(arr: list, target: int) -> int:
    for i in range(len(arr)):
        if arr[i] == target:
            return i

    return -1
```

```python
my_list = [1, 3, 5, 7, 9]
target_value = 5

linear_search(my_list, target_value)
```

### Exercise Two: Adapting a Linear Search algorithm to return the number of occurences of the terget value rather than the first occurence

```python
def linear_search_with_counter(arr: list, target: int) -> int:
    count = 0

    for item in arr:
        if item == target:
            count += 1

    return count
```

```python
my_list = [1, 3, 5, 3, 7, 9, 3]

target_value = 3

linear_search_with_counter(my_list, target_value)
```

### Exercise Three: Implementing a basic recursive implementation of a binary search function

```python
def binary_search_recur(arr: list[int], target: int) -> int:
    middle = len(arr) // 2

    if len(arr) == 0:
        return False

    if len(arr) == 1:
        if arr[middle] == target:
            return middle
        else:
            return False

    if arr[middle] == target:
        return middle
    elif arr[middle] < target:
        result = binary_search_recur(arr[middle+1:], target)
        if result is False:
            return False
        else:
            return middle + 1 + result
    else:
        return binary_search_recur(arr[:middle], target)
```

```python
my_list = [1, 3, 5, 7, 9, 11, 13, 15, 17]

target_value = 13

print(binary_search_recur(my_list, target_value))
```

### Exercise Four: Implementing a basic itertive implementation of a binary search function

```python
def binary_search_iter(arr: list[int], target: int) -> int:
    left = 0
    right = len(arr) - 1

    while left <= right:
        mid_value = (left + right) // 2

        if target == arr[mid_value]:
            return mid_value
        elif target < arr[mid_value]:
            right = mid_value - 1
        else:
            left = mid_value + 1
    return -1
```

```python
my_list = [1, 3, 5, 7, 9, 11, 13, 15, 17]

target_value = 13

binary_search_iter(my_list, target_value)
```

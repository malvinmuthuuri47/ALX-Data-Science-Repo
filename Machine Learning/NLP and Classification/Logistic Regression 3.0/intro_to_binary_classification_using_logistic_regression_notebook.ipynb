```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
# Read data in and view first few entries
df = pd.read_csv('https://raw.githubusercontent.com/Explore-AI/Public-Data/master/Data/classification_sprint/claims_data.csv')
df.head()
```

```python
df['insurance_claim'].unique()
```

### Preprocessing

```python
# labels
y = df['insurance_claim']

# features
X = df.drop('insurance_claim', axis=1)
```

```python
# Transforming the features
X_transformed = pd.get_dummies(X, drop_first=True)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_transformed,
    y,
    test_size=0.2,
    random_state=50
)
```

### Training

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
lr.intercept_[0]
```

```python
lr.coef_
```

```python
lr.coef_.T
```

```python
coeff_df = pd.DataFrame(
    lr.coef_.T,
    X_transformed.columns,
    columns=['Coefficient']
)

coeff_df
```

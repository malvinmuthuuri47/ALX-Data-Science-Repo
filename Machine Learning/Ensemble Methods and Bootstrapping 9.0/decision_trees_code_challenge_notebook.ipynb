```python
population_df = pd.read_csv('https://raw.githubusercontent.com/Explore-AI/Public-Data/master/AnalyseProject/world_population.csv', index_col='Country Code')
population_df.head()
```

### Challenge 1: Population Growth

```python
def get_population_growth_rate_by_country_year(df, country_code):
    country_series = df.loc[country_code]
    
    # create an empty list to store the results
    results = []
    
    for i in range(1, len(country_series)):
        prev_pop = country_series.iloc[i - 1]
        curr_pop = country_series.iloc[i]
        
        year = country_series.index[i]
        
        growth_rate = (curr_pop - prev_pop) / prev_pop
        
        results.append([
            float(year),
            round(growth_rate, 5)
        ])
    
    return np.array(results)
```

```python
get_population_growth_rate_by_country_year(population_df, 'ABW')
```

### Challenge 2: Even-odd train-test split

```python
def feature_response_split(arr):
    X = arr[:, 0]
    y = arr[:, 1]
    
    # split masks
    even_mask = X % 2 == 0
    
    X_train = X[even_mask]
    X_test = X[~even_mask]
    y_train = y[even_mask]
    y_test = y[~even_mask]
    
    return (X_train, y_train), (X_test, y_test)
```

```python
data = get_population_growth_rate_by_country_year(population_df, 'ABW')
(X_train, y_train), (X_test, y_test) = feature_response_split(data)
```

### Challenge 3: Fitting a Model using sklearn's `DecisionTreeRegressor`

```python
X_train = X_train.reshape(-1,1)
X_test = X_test.reshape(-1,1)
```

```python
def train_model(X_train, y_train, MaxDepth):
    X_train = X_train.reshape(-1,1)
    
    tree = DecisionTreeRegressor(max_depth=MaxDepth, random_state=42)
    tree.fit(X_train, y_train)
    
    return tree
```

```python
data = get_population_growth_rate_by_country_year(population_df, 'ABW')
(X_train, y_train), _ = feature_response_split(data)

train_model(X_train, y_train, 3).predict([[2017]])
```

### Challenge 4: Now we would like to test our model on the testing data that we produced in Exercise 2. This test will give the Root Mean Squared Logarithmic Error (RMSLE)

```python
from sklearn.metrics import mean_squared_log_error
```

```python
def test_model(model, y_test, X_test):
    X_test = X_test.reshape(-1, 1)
    y_pred = model.predict(X_test)
    
    y_pred = np.clip(y_pred, 0, None)
    y_test = np.clip(y_test, 0, None)
    
    rmsle = np.sqrt(mean_squared_log_error(y_test, y_pred))
    
    return float(round(rmsle, 3))
```

```python
data = get_population_growth_rate_by_country_year(population_df, 'ABW')
(X_train, y_train), (X_test, y_test) = feature_response_split(data)

lm = train_model(X_train, y_train, 3)
test_model(lm, y_test, X_test)
```

### Task One: Calculating the average GDP for each Region

```python
%%sql

SELECT
    region,
    country_name,
    time_period,
    est_gdp_in_billions,
    AVG(est_gdp_in_billions) OVER (PARTITION BY region) AS Avg_gdp_in_billions
FROM
    access_to_basic_services
WHERE
    time_period = 2020;
```

### Task Two: Filtering the data to include Sub-saharan African Countries with underdeveloped economies and limited access to managed drinking water

```python
%%sql

WITH avg_gdp AS (
    SELECT
        region,
        country_name,
        est_gdp_in_billions,
        pct_managed_drinking_water_services,
        AVG(est_gdp_in_billions) OVER (PARTITION BY REGION) AS avg_gdp_in_billions
    FROM
        access_to_basic_services
    WHERE
        time_period = 2020
        AND
        region = 'Sub-Saharan Africa'
)
SELECT
    region,
    country_name,
    est_gdp_in_billions,
    pct_managed_drinking_water_services
FROM
    avg_gdp
WHERE
    est_gdp_in_billions < avg_gdp_in_billions
    AND
    pct_managed_drinking_water_services < 60;
```

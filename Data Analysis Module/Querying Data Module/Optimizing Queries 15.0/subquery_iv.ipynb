### Task One: A query that displays the average GDP of a country during the Year 2020

```python
%%sql

SELECT
    *
FROM
    economic_indicators
LIMIT 5;
```

```python
%%sql

SELECT
    country_name,
    AVG(est_gdp_in_billions)
FROM
    economic_indicators
WHERE
    time_period = 2020
GROUP BY
    country_name;
```

### Task Two: Joining the Economic indicators and Basic Services Table

```python
%%sql

SELECT
    *
FROM
    basic_services
LIMIT 5;
```

```python
%%sql

SELECT
    eco.country_name,
    eco.time_period,
    eco.est_gdp_in_billions,
    eco.est_population_in_millions,
    eco.pct_unemployment,
    bas.pct_managed_drinking_water_services,
    bas.pct_managed_sanitation_services
FROM
    economic_indicators AS eco
INNER JOIN
    basic_services AS bas
ON
    eco.country_name = bas.country_name
    AND
    eco.time_period = eco.time_period;
```

### Task Three: Using a subquery to determine the results in the WHERE clause

```python
%%sql

SELECT
    eco.country_name,
    eco.time_period,
    eco.est_gdp_in_billions,
    eco.est_population_in_millions,
    eco.pct_unemployment,
    bas.pct_managed_drinking_water_services,
    bas.pct_managed_sanitation_services
FROM
    economic_indicators AS eco
INNER JOIN
    basic_services AS bas
ON
    eco.country_name = bas.country_name
    AND
    eco.time_period = eco.time_period
WHERE
    eco.time_period = 2020
    AND
    bas.pct_managed_drinking_water_services < 90
    AND
    eco.est_gdp_in_billions > (
        SELECT
            AVG(est_gdp_in_billions)
        FROM
            economic_indicators
        WHERE
            time_period = 2020
    )
```

### Task One: Calculating the Total Land Area of the Middle Africa Subregion

```python
%%sql

SELECT
    SUM(Land_area) AS Total_land_area
FROM
    geographic_location
WHERE
    Sub_region = 'Middle Africa';
```

### Task Two: Calculating Land Area Percentages for the Middle African countries using a static value

```python
%%sql

SELECT
    Country_name,
    Sub_region,
    ROUND((land_area / 3888270.00) * 100) AS pct_regional_land
FROM
    geographic_location
WHERE
    Sub_region = 'Middle Africa';
```

### Task Three: Calculating the Land Percentages using a subquery instead of a static value

```python
%%sql

SELECT
    Country_name,
    Sub_region,
    land_area,
    (
        Round((land_area / (
            SELECT
                SUM(Land_area)
            FROM
                geographic_location
            WHERE
                Sub_region = 'Middle Africa'
        ) * 100))
    ) AS pct_regional_land
FROM
     geographic_location
WHERE
    Sub_region = 'Middle Africa';
```

### Task Four: Using correlated subqueries to calculate the land percentages for all the regions

```python
%%sql

SELECT
    Country_name,
    Sub_region,
    Land_area,
    Round((land_area / (
        SELECT
            SUM(land_area)
        FROM
            geographic_location
        WHERE
            g.sub_region = Sub_region
    ) * 100)) AS pct_regional_land
FROM
    geographic_location AS g;
```

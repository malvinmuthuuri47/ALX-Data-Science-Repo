### Task One: Creating a water_level_assessments table

```python
%%sql

CREATE TABLE water_level_assessments (
    Assessment_id INTEGER PRIMARY KEY AUTOINCREMENT,
    year TEXT,
    Officer_Reg TEXT,
    water_level TEXT
)
```

```python
%%sql

INSERT INTO water_level_assessments (
    assessment_id,
    year,
    officer_reg,
    water_level
)
SELECT
    assessment_id,
    year,
    officer_reg,
    water_level
FROM
    assessments;
```

```python
%%sql

SELECT
    *
FROM
    water_level_assessments
ORDER BY
    YEAR;
```

### Task Two: Creating an officer's table

```python
%%sql

CREATE TABLE officers (
    officer_reg TEXT PRIMARY KEY,
    officer_name TEXT
)
```

```python
%%sql

INSERT INTO officers (
    officer_reg,
    officer_name
)
SELECT
    officer_reg,
    assessment_officer
FROM
    assessments
GROUP BY
    assessment_officer;
```

```python
%%sql

SELECT
    *
FROM
    officers;
```

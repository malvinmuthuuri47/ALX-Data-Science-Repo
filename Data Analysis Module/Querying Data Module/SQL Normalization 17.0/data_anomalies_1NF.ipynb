### Task One: Creating a 1NF compliant employees table

```python
%%sql

DROP TABLE IF EXISTS Employees_1NF;

CREATE TABLE Employees_1NF (
	Name VARCHAR NOT NULL,
	Surname VARCHAR NOT NULL,
	Department VARCHAR NOT NULL,
	Role VARCHAR NOT NULL,
	Title VARCHAR,
	OccupationBand VARCHAR,
	Salary REAL,
	PRIMARY KEY (Name, Surname, Department, Role)
);
```

### Task Two: Splitting Fullname into Name and Surname columns

```python
%%sql

SELECT
    fullname,
    TRIM(SUBSTR(fullname, 1, INSTR(fullname, ',') -1)) AS Name,
    TRIM(SUBSTR(fullname, INSTR(fullname, ',') + 1)) AS Surname,
    UPPER(SUBSTR(Title,1,1)) || LOWER(SUBSTR(Title, 2)) AS Title
FROM
    employees;
```

### Task Three: Splitting non-atomic values in Role and Department columns

```python
%%sql

SELECT
    TRIM(SUBSTR(fullname, 1, INSTR(fullname, ',') -1)) AS Name,
    TRIM(SUBSTR(fullname, INSTR(fullname, ',') + 1)) AS Surname,
    UPPER(SUBSTR(Title,1,1)) || LOWER(SUBSTR(Title, 2)) AS Title,
    Role,
    OccupationBand,
    Salary,
    Department
FROM
    employees
WHERE
    role LIKE '%,%'
    OR
    Department LIKE '%,%';
```

##### Subtask01: Returning the First role and Department in non-atomic columns in the table

```python
%%sql

SELECT
	TRIM(SUBSTR(Fullname, 1, INSTR(FullName, ',')-1)) AS Name,
	TRIM(SUBSTR(Fullname, INSTR(FullName, ',')+1)) AS Surname,
	UPPER(SUBSTR(Title,1,1)) || LOWER(SUBSTR(Title, 2)) AS Title,

	CASE
		WHEN
			TRIM(SUBSTR(Role,1,INSTR(Role,',')-1)) = ''
		THEN
			Role
		ELSE
			TRIM(SUBSTR(Role,1,INSTR(Role,',')-1))
	END AS Role,
	OccupationBand,
	Salary,

	CASE
		WHEN
			TRIM(SUBSTR(Department,1,INSTR(Department,',')-1))=''
		THEN
			Department
		ELSE
			TRIM(SUBSTR(Department,1,INSTR(Department,',')-1))
	END AS Department
FROM
	Employees
WHERE
	Role LIKE '%,%'
	OR
	Department LIKE '%,%'
```

##### Subtask02: Returning the Second role and Department in non-atomic columns in the table

```python
%%sql

SELECT
	TRIM(SUBSTR(Fullname, 1, INSTR(FullName, ',')-1)) AS Name,
	TRIM(SUBSTR(Fullname, INSTR(FullName, ',')+1)) AS Surname,
	UPPER(SUBSTR(Title,1,1)) || LOWER(SUBSTR(Title, 2)) AS Title,

	CASE
		WHEN
			TRIM(SUBSTR(Role,1,INSTR(Role,',')-1)) = ''
		THEN
			Role
		ELSE
			TRIM(SUBSTR(Role,INSTR(Role,',')+1))
	END AS Role,
	OccupationBand,
	Salary,

	CASE
		WHEN
			TRIM(SUBSTR(Department,1,INSTR(Department,',')-1))=''
		THEN
			Department
		ELSE
			TRIM(SUBSTR(Department,INSTR(Department,',')+1))
	END AS Department
FROM
	Employees
WHERE
	Role LIKE '%,%'
	OR
	Department LIKE '%,%'
```

### Task Four: Combining all the atomic columns into one table

```python
%%sql

SELECT
	TRIM(SUBSTR(Fullname, 1, INSTR(FullName, ',')-1)) AS Name,
	TRIM(SUBSTR(Fullname, INSTR(FullName, ',')+1)) AS Surname,
	UPPER(SUBSTR(Title,1,1)) || LOWER(SUBSTR(Title, 2)) AS Title,

	CASE
		WHEN
			TRIM(SUBSTR(Role,1,INSTR(Role,',')-1)) = ''
		THEN
			Role
		ELSE
			TRIM(SUBSTR(Role,1,INSTR(Role,',')-1))
	END AS Role,
	OccupationBand,
	Salary,

	CASE
		WHEN
			TRIM(SUBSTR(Department,1,INSTR(Department,',')-1))=''
		THEN
			Department
		ELSE
			TRIM(SUBSTR(Department,1,INSTR(Department,',')-1))
	END AS Department
FROM
	Employees
WHERE
	Role LIKE '%,%'
	OR
	Department LIKE '%,%'

UNION

SELECT
	TRIM(SUBSTR(Fullname, 1, INSTR(FullName, ',')-1)) AS Name,
	TRIM(SUBSTR(Fullname, INSTR(FullName, ',')+1)) AS Surname,
	UPPER(SUBSTR(Title,1,1)) || LOWER(SUBSTR(Title, 2)) AS Title,

	CASE
		WHEN
			TRIM(SUBSTR(Role,1,INSTR(Role,',')-1)) = ''
		THEN
			Role
		ELSE
			TRIM(SUBSTR(Role,INSTR(Role,',')+1))
	END AS Role,
	OccupationBand,
	Salary,

	CASE
		WHEN
			TRIM(SUBSTR(Department,1,INSTR(Department,',')-1))=''
		THEN
			Department
		ELSE
			TRIM(SUBSTR(Department,INSTR(Department,',')+1))
	END AS Department
FROM
	Employees
WHERE
	Role LIKE '%,%'
	OR
	Department LIKE '%,%'

UNION

SELECT
	TRIM(SUBSTR(Fullname, 1, INSTR(FullName, ',')-1)) AS Name,
	TRIM(SUBSTR(Fullname, INSTR(FullName, ',')+1)) AS Surname,
	UPPER(SUBSTR(Title,1,1)) || LOWER(SUBSTR(Title, 2)) AS Title,
    Role,
	OccupationBand,
	Salary,
    Department
FROM
	Employees
WHERE
	Role NOT LIKE '%,%'
	OR
	Department NOT LIKE '%,%';
```

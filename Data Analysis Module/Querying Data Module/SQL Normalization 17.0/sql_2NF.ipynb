### Splitting the company_employees table to ensure it is in the 2NF

```python
%%sql

CREATE TABLE employees AS
SELECT
    DISTINCT employee_id,
    name,
    state_code,
    home_state
FROM
    company_employees;
```

```python
%sql SHOW TABLES;
```

```python
%%sql

SELECT
    *
FROM
    employees;
```

```python
%%sql

CREATE TABLE jobs AS
SELECT
    DISTINCT job_code,
    job_title
FROM
    company_employees;
```

```python
%sql SHOW TABLES;
```

```python
%%sql

SELECT
    *
FROM
    jobs;
```

```python
%%sql

ALTER TABLE
    jobs
ADD
    PRIMARY KEY (job_code);
```

```python
%%sql

ALTER TABLE
    employees
ADD
    PRIMARY KEY (employee_id);
```

```python
%sql SHOW TABLES;
```

```python
%%sql

SELECT
    *
FROM
    employees;
```

```python
%%sql

SELECT
    *
FROM
    jobs;
```

```python
%%sql

CREATE TABLE employee_roles AS
SELECT DISTINCT
    employee_id,
    job_code
FROM
    company_employees;
```

```python
%sql SHOW TABLES;
```

```python
%%sql

SELECT
    *
FROM
    employee_roles;
```

```python
%sql DROP TABLE employee_roles;
```

```python
%%sql

CREATE TABLE employee_roles (PRIMARY KEY(employee_id, job_code)) AS
SELECT
    employee_id,
    job_code
FROM
    company_employees;
```

```python
%%sql

SELECT
    *
FROM
    employee_roles;
```

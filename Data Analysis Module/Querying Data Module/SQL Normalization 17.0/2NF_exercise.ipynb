### Task One: Creating the assessments table

```python
%sql PRAGMA table_info(dam_levels_inf);
```

```python
%%sql

CREATE TABLE assessments (
    AssessmentId INTEGER PRIMARY KEY,
    year TEXT,
    Assessment_officer TEXT,
    Officer_Reg TEXT,
    water_level TEXT
);
```

```python
%%sql

INSERT INTO assessments (
    assessmentid,
    year,
    assessment_officer,
    officer_reg,
    water_level
)
SELECT
    id,
    year,
    assessment_officer,
    officer_reg,
    water_level
FROM
    dam_levels_inf;
```

```python
%%sql

SELECT
    *
FROM
    assessments;
```

```python
%%sql

ALTER TABLE assessments
RENAME COLUMN AssessmentId TO Assessment_Id;
```

```python
%%sql

SELECT
    *
FROM
    assessments;
```

### Task Two: Creating the Dams table

```python
%%sql

CREATE TABLE dams (
    dam_name TEXT,
    dam_latitude TEXT,
    dam_longitude TEXT
);
```

```python
%%sql

INSERT INTO dams (
    dam_name,
    dam_latitude,
    dam_longitude
)
SELECT
    dam_name,
    dam_latitude,
    dam_longitude
FROM
    dam_levels_inf;
```

```python
%%sql

SELECT
    *
FROM
    dam_levels_inf
GROUP BY
    dam_name;
```

```python
%sql DROP TABLE dams;
```

### Task Three: Creating a junction table

```python
%%sql

CREATE TABLE dam_assessments (
    assessment_id INTEGER,
    dam_name TEXT,
    PRIMARY KEY(assessment_id, dam_name),
    FOREIGN KEY (assessment_id) REFERENCES assessments(assessment_id),
    FOREIGN KEY (dam_name) REFERENCES dams(dam_name)
);
```

```python
%sql PRAGMA table_list;
```

```python
%sql PRAGMA table_info(dam_assessments);
```

```python
%%sql

SELECT
    *
FROM
    dam_assessments;
```

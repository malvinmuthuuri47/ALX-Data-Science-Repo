### Task Two: A query that shows the age of all employees when they were hired

```python
%%sql

SELECT
    firstname,
    lastname,
    ROUND((julianday(hiredate) - julianday(birthdate)) / 365.25, 0)  AS Age_whe_hired
FROM
    employees
LIMIT 5;
```

### Task Three: Calculating the month-to-month revenue at chinook

```python
%%sql

SELECT
    *
FROM
    invoices
LIMIT 5;
```

```python
%%sql

SELECT
    SUBSTR(invoicedata, 1, 7) AS "Month",
    SUM(total) AS "Revenue"
FROM
    invoices
GROUP BY
    InvoiceID
ORDER BY
    INVOICEID
LIMIT 5;
```

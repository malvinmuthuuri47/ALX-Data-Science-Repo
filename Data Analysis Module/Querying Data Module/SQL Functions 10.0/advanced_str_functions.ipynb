### Task One: Converting the data type of Employees salaries into string data types

##### Step One: Describing the tables and listing the rows in the tables

```python
%%sql

SELECT
    name, sql
FROM
    sqlite_master
WHERE
    type = 'table';
```

```python
%sql PRAGMA table_info(employees)
```

```python
%%sql

SELECT
    *,
    CAST(SALARY AS STRING) AS CAST_SALARY
FROM
    employees
LIMIT 10;
```

### Task Two: Trimming leading and trailing white spaces from the Shipname Column

```python
%%sql

SELECT
    *
FROM
    ORDERS
LIMIT 5;
```

```python
%%sql

SELECT
    *,
    TRIM(shipname) AS Trimmed_shipname
FROM
    orders
LIMIT 5;
```

### Task Three: Replacing Strings with alternative values

```python
%%sql

SELECT
    *,
    IFNULL(contacttitle, 'N/A') AS New_contact_Title
FROM
    customers;
```

### Task Four: Writing a query that extracts the first three characters from the City column in the customer's table

```python
%%sql

SELECT
    *,
    SUBSTRING(city, 1, 3) AS New_city_Column
FROM
    customers;
```

### Task Five: Concatenating the firstname and lastname of employees in the Employees table into a column called fullname

```python
%%sql

SELECT
    *,
    CONCAT(firstname, ' ', lastname) AS FullName
FROM
    employees
LIMIT 5;
```

### Task One: Extracting Country names without information inside the parenthesis

##### Step One: Retrieving distinct country names from the Database table

```python
%%sql

SELECT
    DISTINCT country_name
FROM
    access_to_basic_services
WHERE
    country_name LIKE '%(%)%';
```

##### Step Two: Extracting the country names without the brackets

```python
%%sql

SELECT
    DISTINCT country_name,
    LEFT(country_name, POSITION('(' IN country_name) - 1) AS New_country_name,
    LENGTH(LEFT(country_name, POSITION('(' IN country_name) - 1)) AS New_country_name_length
FROM
    access_to_basic_services
WHERE
    country_name LIKE '%(%)%';
```

### Task Two: Identifying any extra characters such as spaces and getting rid of them

```python
%%sql

SELECT
    DISTINCT country_name,
    TRIM(LEFT(country_name, POSITION('(' IN country_name) - 1)) AS New_country_name,
    LENGTH(TRIM(LEFT(country_name, POSITION('(' IN country_name) - 1))) AS New_trimmed_country_name_length
FROM
    access_to_basic_services
WHERE
    country_name LIKE '%(%)%';
```

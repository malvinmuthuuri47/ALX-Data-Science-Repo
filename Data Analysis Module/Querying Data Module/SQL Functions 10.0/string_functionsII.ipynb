### Task One: Viewing the Datatypes of the customer's table

```python
%%sql

SELECT
    sql
FROM
    sqlite_master
WHERE
    type='table'
    AND
    name='customers';
```

### Task Two: A query that returns all the columns for the tracks that contain the word love

```python
%%sql

SELECT
    *
FROM
    tracks
WHERE
    name LIKE '%LOVE%'
LIMIT 10;
```

### Task Three: A query that returns all customers whose first name starts with the letter D

```python
%%sql

SELECT
    *
FROM
    customers
WHERE
    firstName LIKE 'D%';
```

### Task Four: A query that returns customers with email addresses that have domains ending with three letters

```python
%%sql

SELECT
    *
FROM
    customers
WHERE
    SUBSTRING(email, -3) GLOB '[A-Za-z][A-Za-z][A-Za-z]';
```

### Task Five: A query that returns all customers living in the UK

```python
%%sql

SELECT
    *
FROM
    customers
WHERE
    country LIKE '%United Kingdom%';
```

### Task Six: A query that returns an employee's first and last names in the same column

```python
%%sql

SELECT
    CONCAT(firstName, ' ', lastname) AS Employee_Full_name
FROM
    employees
LIMIT 5;
```

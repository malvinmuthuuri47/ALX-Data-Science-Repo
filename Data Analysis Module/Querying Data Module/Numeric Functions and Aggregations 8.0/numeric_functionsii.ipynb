### Task One: Calculate the GDP per Year for each Country

```python
%%sql

SELECT
    country_name,
    time_period,
    est_gdp_in_billions
FROM
    access_to_basic_services
LIMIT 5;
```

### Task Two: Calclate the GDP per Year for each Country, rounding off the Gdp_in_billions column

```python
%%sql

SELECT
    country_name,
    time_period,
    Round(est_gdp_in_billions)
FROM
    access_to_basic_services;
```

### Task Three: Calculate the Logarithm of GDP for each country over the time period

```python
%%sql

SELECT
    country_name,
    time_period,
    Log(est_gdp_in_billions)
FROM
    access_to_basic_services
Limit 5;
```

### Task Four: Calculate the Square root of the GDP for each country over the time period

```python
%%sql

SELECT
    country_name,
    time_period,
    SQRT(est_gdp_in_billions)
FROM
    access_to_basic_services
LIMIT 5;
```

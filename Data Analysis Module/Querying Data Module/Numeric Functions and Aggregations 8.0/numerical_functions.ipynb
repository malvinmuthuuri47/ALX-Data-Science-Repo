### Task One: What is the total number of entries in the dataset

```python
%%sql

SELECT
    COUNT(*) AS Total_number_of_entries
FROM
    access_to_basic_services;
```

### Task Two: Determine the earliest and latest years for which we have data in the dataset

```python
%%sql

SELECT
    MIN(Time_period) AS earliest_year_in_dataset,
    MAX(Time_period) AS latest_year_in_dataset
FROM
    access_to_basic_services;
```

### Task Three: Determining the total number of countries included in the dataset

```python
%%sql

SELECT
    COUNT(DISTINCT country_name) AS Distinct_countries_in_dataset
FROM
    access_to_basic_services;
```

### Task Four: Average percentage of people with access to managed drinking water services across all years and all countries in the dataset

```python
%%sql

SELECT
    Time_period,
    country_name,
    AVG(pct_managed_drinking_water_services) AS Average_percentage_of_people_with_access_to_managed_drinking_water_services
FROM
    access_to_basic_services
GROUP BY
    Time_period, country_name;
```

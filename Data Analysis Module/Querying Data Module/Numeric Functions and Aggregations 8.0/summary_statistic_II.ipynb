##### SQL Code to build upon

```python
%%sql

SELECT
    region,
    sub_region,
    MIN(pct_managed_drinking_water_services),
    MAX(pct_managed_drinking_water_services),
    AVG(pct_managed_drinking_water_services),
    COUNT(DISTINCT(country_name)) AS Number_of_countries,
    SUM(est_gdp_in_billions) AS Total_GDP
FROM
    access_to_basic_services
GROUP BY
    region, sub_region
ORDER BY
    Total_GDP;
```

### Task One: Build upon the base SQL query and filter the results for the time period that is 2020

```python
%%sql

SELECT
    region,
    sub_region,
    Time_period,
    MIN(pct_managed_drinking_water_services),
    MAX(pct_managed_drinking_water_services),
    AVG(pct_managed_drinking_water_services),
    COUNT(DISTINCT(country_name)) AS Number_of_countries,
    SUM(est_gdp_in_billions) AS Total_GDP
FROM
    access_to_basic_services
WHERE
    Time_period = 2020
GROUP BY
    region, sub_region
ORDER BY
    Total_GDP;
```

### Task Two: Return results for countries whose percentage of managed drinking water services is below 60% using the Where Clause

```python
%%sql

SELECT
    region,
    sub_region,
    Time_period,
    MIN(pct_managed_drinking_water_services) AS MIN_percentage_water_access,
    MAX(pct_managed_drinking_water_services) AS MAX_percentage_water_access,
    AVG(pct_managed_drinking_water_services) AS AVG_percentage_water_access,
    COUNT(DISTINCT(country_name)) AS Number_of_countries,
    SUM(est_gdp_in_billions) AS Total_GDP
FROM
    access_to_basic_services
WHERE
    Time_period = 2020
    AND
    pct_managed_drinking_water_services < 60
GROUP BY
    region, sub_region
ORDER BY
    Total_GDP;
```

### Task Three: Filtering regions and sub_regions that have fewer than four countries in the Number_of_countries alias

```python
%%sql

SELECT
    region,
    sub_region,
    Time_period,
    MIN(pct_managed_drinking_water_services) AS MIN_percentage_water_access,
    MAX(pct_managed_drinking_water_services) AS MAX_percentage_water_access,
    AVG(pct_managed_drinking_water_services) AS AVG_percentage_water_access,
    COUNT(DISTINCT(country_name)) AS Number_of_countries,
    SUM(est_gdp_in_billions) AS Total_GDP
FROM
    access_to_basic_services
WHERE
    Time_period = 2020
    AND
    pct_managed_drinking_water_services < 60
GROUP BY
    region, sub_region
HAVING
    Number_of_countries < 4
ORDER BY
    Total_GDP;
```

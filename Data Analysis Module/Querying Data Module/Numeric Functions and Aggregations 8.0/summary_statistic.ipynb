### Task One: Determining the Maximum, Minimum, and Average percentage of people with access to managed drinking water services per region and sub region

```python
%%sql

SELECT
    region,
    sub_region,
    MAX(pct_managed_drinking_water_services) AS Max_percentage_with_access_to_drinking_water,
    MIN(pct_managed_drinking_water_services) AS Min_percentage_with_access_to_drinking_water,
    AVG(pct_managed_drinking_water_services) AS Average_percentage_with_access_to_drinking_water
FROM
    access_to_basic_services
GROUP BY
    region,
    sub_region;
```

### Task Two: Total number of countries within each region and sub region

```python
%%sql

SELECT
    region,
    sub_region,
    COUNT(country_name)
FROM
    access_to_basic_services
GROUP BY
    region,
    sub_region;
```

### Task Three: Calculating the Total GDP for each region and sub region

```python
%%sql

SELECT
    region,
    sub_region,
    SUM(est_gdp_in_billions) AS Total_GDP
FROM
    access_to_basic_services
GROUP BY
    region,
    sub_region;
```

### Task Two: Using the OrderDetails table, show the total quantity of products sold for each ProductID

##### Step One: List some few rows from the OrderDetails Table

```python
%%sql

SELECT
    *
FROM
    orderDetails
LIMIT 5;
```

```python
%%sql

SELECT
    ProductID,
    COUNT(quantity) AS Total_Quantity_Per_ProductID
FROM
    OrderDetails
GROUP BY
    ProductID;
```

### Task Three: Find the Maximum, Minimum, and average freight from the orders table 

##### Step One: List some few rows from the Orders table;

```python
%%sql

SELECT
    *
FROM
    ORDERS;
```

```python
%%sql

SELECT
    MIN(freight) AS Min_Freight,
    MAX(freight) AS Max_Freight,
    AVG(freight) AS Average_Freight
FROM
    orders
LIMIT 5;
```

### Task Four: Number of unique customers that have made orders from the orders table

```python
%%sql

SELECT
    COUNT(DISTINCT CustomerID) AS Unique_Customer_Entries
FROM
    orders;
```

### Task Five: Calculating the total price for each order

##### Step One: List some rows from the OrderDetails table

```python
%%sql

SELECT
    *
FROM
    OrderDetails
LIMIT 5;
```

```python
%%sql

SELECT
    OrderID,
    ROUND(SUM(UnitPrice * quantity - Discount), 2) AS Total_Price
FROM
    OrderDetails
GROUP BY
    OrderID
ORDER BY
    OrderID;
```

### Task Six: Finding the 5 Largest Orders processed

##### Step One: List some few rows from the OrderDetails Table

```python
%%sql

SELECT
    *
FROM
    OrderDetails
LIMIT 5;
```

##### Step Two: Write the SQL Code

```python
%%sql

SELECT
    ProductID,
    ROUND(SUM(UnitPrice * Quantity - Discount)) AS Total_Price_Per_ProductID
FROM
    OrderDetails
GROUP BY
    ProductID
ORDER BY
    Total_Price_Per_ProductID DESC;
```

### Task Four: A SQL Query that returns all customers living in Germany

```python
%%sql

SELECT
    FirstName, LastName
FROM
    customers
WHERE
    Country = 'Germany'
LIMIT 5;
```

### Task Five: A SQL Query returning the First name, last name, and Country columns of the customer table for those that don't live in Germany

```python
%%sql

SELECT
    FirstName, LastName, Country
FROM
    customers
WHERE
    Country <> 'Germany'
LIMIT 10;
```

### Task Six: A Query that returns all invoices that show USA purchases with a total greater than 10 dollars

```python
%%sql

SELECT
    *
FROM
    invoices
WHERE
    (
        billingcountry = 'USA'
        AND
        total > 10
    );
```

### Task Seven: A query that returns all stored information for Sales Support agents hired on or after 3rd May 2003 residing in Calgary

```python
%%sql

SELECT
    *
FROM
    employees
WHERE
    (
        hiredate >= 2003-05-03
        AND
        title = 'Sales Support Agent'
        AND
        city = 'Calgary'
    );
```

### Task One: Finding the names of all customers from Germany

```python
%%sql

SELECT
    *
FROM
    customers
LIMIT 5;
```

```python
%%sql

SELECT
    *
FROM
    customers
WHERE
    country LIKE '%Germany%';
```

### Task Two: Finding all products in the Northwind database whose unit price > 30 and are discontinued

##### Step One: Listing the content of the Products table

```python
%%sql

SELECT
    *
FROM
    products
LIMIT 5;
```

##### Step Two: Answering the question

```python
%%sql

SELECT
    *
FROM
    products
WHERE
    (
        unitprice > 30
        AND
        discontinued = 1
    );
```

### Task Three: Finding the contact names and phone numbers of customers from USA or Canada

##### Step One: Show data from the customers table

```python
%%sql

SELECT
    *
FROM
    customers
LIMIT 5;
```

##### Step Two: Find contact names and phone numbers of customers from USA or Canada

```python
%%sql

SELECT
    contactname,
    phone
FROM
    customers
WHERE
    country IN ('USA', 'Canada');
```

### Task Four: Select all orders placed by VINET, SUPRD, or Quick that were shipped via shipper 1 or 2

##### Step One: Retrieve data from the orders table before answering the question

```python
%%sql

SELECT
    *
FROM
    orders
LIMIT 5;
```

##### Step Two: Answer the question

```python
%%sql

SELECT
    *
FROM
    orders
WHERE
    customerid IN ('VINET', 'QUICK', 'SUPRD')
    AND
    shipvia IN (1, 2);
```

### Task Five: Finding Employees in the Northwind Database who are either Sales reps or Sales Managers based in London

##### Step One: Retrieving data from the Employees table

```python
%%sql

SELECT
    *
FROM
    employees
LIMIT 5;
```

##### Step Two: Answer the question

```python
%%sql

SELECT
    *
FROM
    employees
WHERE
    city LIKE 'London'
    AND
    title IN ('Sales Representative', 'Sales Manager')
```

### Task One: Selecting the data required for the Analysis

```python
%%sql

SELECT
    sub_region,
    country_name,
    land_area
FROM
    access_to_basic_services
WHERE
    Time_period = 2020
    AND
    Land_area IS NOT NULL;
```

### Task Two: Calculating the land area covered as a percentage of the Country's Subregion

```python
%%sql

SELECT
    sub_region,
    country_name,
    Round(land_area * 100 / SUM(land_area) OVER (PARTITION BY sub_region), 2) AS pct_sub_region_land_area
FROM
    access_to_basic_services
WHERE
    land_area IS NOT NULL;
```

### Task Three: Calculating the running population average for each country's subregion

```python
%%sql

SELECT
    sub_region,
    country_name,
    time_period,
    pct_managed_drinking_water_services,
    pct_managed_sanitation_services,
    est_gdp_in_billions,
    ROUND(AVG(est_population_in_millions) OVER (PARTITION BY sub_region ORDER BY sub_region), 2) AS Running_Average_Population
FROM
    access_to_basic_services
WHERE
    est_gdp_in_billions IS NOT NULL;
```

### Base Query

```python
%%sql

SELECT
    country_name,
    time_period,
    pct_managed_drinking_water_services
FROM
    access_to_basic_services;
```

### Task One: Ordering countries based on their water access levels per year

```python
%%sql

SELECT
    country_name,
    time_period,
    ROW_NUMBER() OVER (PARTITION BY time_period ORDER BY pct_managed_drinking_water_services) AS Percentage_Yearly_Water_Access_level
FROM
    access_to_basic_services;
```

### Task Two: Assessing the Rankings for countries with the same water level access in a given year

```python
%%sql

SELECT
    country_name,
    time_period,
    pct_managed_drinking_water_services,
    ROW_NUMBER() OVER (PARTITION BY time_period ORDER BY pct_managed_drinking_water_services) AS Percentage_Yearly_Water_Access_level
FROM
    access_to_basic_services
WHERE
    pct_managed_drinking_water_services = 100;
```

### Task Three: Replacing the Row_NUMBER() with Rank() function to see the difference

```python
%%sql

SELECT
    country_name,
    time_period,
    pct_managed_drinking_water_services,
    RANK() OVER (PARTITION BY time_period ORDER BY pct_managed_drinking_water_services) AS Percentage_Yearly_Water_Access_level
FROM
    access_to_basic_services;
```

### Task One: Determining the list of products and their suppliers

```python
%%sql

SELECT
    prd.productName,
    sup.companyname
FROM
    products AS prd
INNER JOIN
    suppliers AS sup
ON
    prd.supplierid = sup.supplierid;
```

### Task Two: Generating a list of all cities where customers and employees are located

```python
%%sql

SELECT
    city
FROM
    customers
UNION
SELECT
    city
FROM
    employees;
```

### Task Three: Creating a list of unique cities where customers are located and employees live

```python
%%sql

SELECT
    cstm.city
FROM
    customers AS cstm
INNER JOIN
    employees AS emp
ON
    cstm.city = emp.city;
```

### Task Four: A last quarter report on sales

```python
%%sql

SELECT
    customer.contactname,
    customer.country,
    SUM(orderdetails.unitprice * orderdetails.quantity) AS TotalAmountSpent,
    products.ProductName,
    SUM(orderDetails.quantity) AS TotalPurchasePerProduct
FROM
    Customers AS customer
INNER JOIN
    orders AS orders
ON
    customer.customerid = orders.customerid
INNER JOIN
    orderdetails AS orderdetails
ON
    orders.orderid = orderdetails.orderid
INNER JOIN
    Products AS products
ON
    orderdetails.productid = products.productid
WHERE
    orders.orderdate BETWEEN '1996-08-01' AND '1998-01-30'
GROUP BY
    customer.contactname,
    customer.country,
    products.productname
HAVING
    TotalAmountSpent > 5000;
```

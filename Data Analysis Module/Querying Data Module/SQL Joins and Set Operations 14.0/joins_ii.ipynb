### Task One: Returning The Title of the tracks that match the ablum on the album ID column

```python
%%sql

SELECT
    alb.albumid,
    alb.title,
    tk.name
FROM
    albums AS alb
INNER JOIN
    tracks AS tk
ON
    alb.title = tk.name;
```

### Task Two: Adding an artists column to the previous query

```python
%%sql

SELECT
    alb.albumid,
    alb.title,
    tk.name,
    tk.composer
FROM
    albums AS alb
INNER JOIN
    tracks AS tk
ON
    alb.title = tk.name;
```

### Task Three: Using LEFT JOIN to find items that aren't in any invoice

```python
%%sql

SELECT
    trk.trackid,
    inv.invoiceid
FROM
    tracks AS trk
LEFT JOIN
    invoice_items AS inv
ON
    trk.trackid = inv.trackid
LIMIT 5;
```

### Task Four: Filtering the results from Task Three to those with None

```python
%%sql

SELECT
    trk.trackid,
    inv.invoiceid
FROM
    tracks AS trk
LEFT JOIN
    invoice_items AS inv
ON
    trk.trackid = inv.trackid
WHERE
    inv.invoiceID IS NULL;
```

### Task Five: Listing all possible combinations of media type and Genres tables

```python
%%sql

SELECT
    g.name AS Genre,
    m.name AS MediaType
FROM
    genres AS g
CROSS JOIN
    media_types AS m;
```

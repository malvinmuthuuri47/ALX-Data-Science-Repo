### Task One: Creating a new Table named Geographic_Location

```python
%%sql

CREATE TABLE Geographic_Location (
    Country_name varchar(37) PRIMARY KEY,
    Sub_region varchar(25),
    Region varchar(32),
    Land_area decimal(10,2)
);
```

##### Confirming whether the Geographic_Location table was created

```python
%sql SHOW TABLES;
```

```python
%sql DESCRIBE geographic_location;
```

### Task Two: Extracting Columns and Data from access_to_basic_services and feeding it into the geographic_location table

```python
%%sql

INSERT INTO geographic_location (
    country_name,
    sub_region,
    region,
    land_area
)
SELECT
    country_name,
    sub_region,
    region,
    AVG(land_area) AS Average_Land_Area
FROM
    access_to_basic_services
GROUP BY
    country_name,
    sub_region,
    region;
```

##### Confirming that the data from the access_to_basic_services has been entered into geographic_location

```python
%%sql

SELECT
    *
FROM
    geographic_location
LIMIT 10;
```

### Task One: Creating a table that contains data about access to basic services for each Country and Year only

```python
%%sql

CREATE TABLE Basic_services (
    country_name varchar(37),
    time_period int,
    pct_managed_drinking_water_services decimal(5, 2),
    pct_managed_sanitation_services decimal(5, 2),
    PRIMARY KEY (country_name, time_period),
    FOREIGN KEY (country_name) REFERENCES geographic_location(country_name)
);
```

##### Confirming that the Table Basic_services was successfully created

```python
%sql SHOW TABLES;
```

##### Extracting values from access_to_basic_services into basic_services

```python
%%sql

INSERT INTO basic_services (
    country_name,
    time_period,
    pct_managed_drinking_water_services,
    pct_managed_sanitation_services
)
SELECT
    country_name,
    time_period,
    pct_managed_drinking_water_services,
    pct_managed_sanitation_services
FROM
    access_to_basic_services;
```

##### Confirming that the values have been entered correctly

```python
%%sql

SELECT
    *
FROM
    basic_services
LIMIT 10;
```

### Task Two: Creating a Table called Economic Indicators that contains economic indicators for each Location

```python
%%sql

CREATE TABLE economic_indicators (
    country_name varchar(37),
    time_period int,
    est_gdp_in_billions decimal(8, 2),
    est_population_in_millions decimal(11, 6),
    pct_unemployment decimal(5, 2)
);
```

```python
%%sql

INSERT INTO economic_indicators (
    country_name,
    time_period,
    est_gdp_in_billions,
    est_population_in_millions,
    pct_unemployment
)
SELECT
    country_name,
    time_period,
    est_gdp_in_billions,
    est_population_in_millions,
    pct_unemployment
FROM
    access_to_basic_services;
```

##### Confirming that the table economic_indicators was successfully created

```python
%%sql

SELECT
    *
FROM
    economic_indicators
LIMIT 10;
```

##### Altering the structure of the Economic_indicators table to include Primary and Foreign Keys

```python
%%sql

ALTER TABLE economic_indicators
ADD PRIMARY KEY (country_name, time_period),
ADD FOREIGN KEY (country_name) REFERENCES geographic_location(country_name);
```
